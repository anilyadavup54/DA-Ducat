# Introduction of numpy
Numpy which stands for Numerical Python Numpy in extrimaly popular python module. It is a library consisting of multidimensional array obj and a collection of routinesfor processing of array> It is heavily used in scientic computing. Using numpy mathemetical and logical operations on array can be performed in 2005 Travis Oliphant created Numpy package by incorporating the features of Numarray into Numeric package

```python
import numpy as np
import sys
import time
```

### Why we use numpy array & Diff B/W List and numpy
1. Less Momory
2. Fast Execution
3. Conviniant

```python
# Less Memory
li=range(1000)
print(sys.getsizeof(1)*len(li))#It's print memory size that consume - 28 byte

arr=np.arange(1000)
print(arr.size*arr.itemsize) #It's print memory size that consume -8 byte
```

```python
# Fast execution
numbers = list(range(1_000_000))

start = time.perf_counter()
list_result = [number + 10 for number in numbers]
list_time = time.perf_counter() - start

numpy_numbers = np.arange(1_000_000)
start = time.perf_counter()
numpy_result = numpy_numbers + 10
numpy_time = time.perf_counter() - start

print(f"Python list time: {1000*list_time:.2f} seconds")
print(f"NumPy array time: {1000*numpy_time:.2f} seconds") # NUMPY 10X FASTER THEN LIST
print(f"First five results: {numpy_result[:5]}")
```

```python
# Convenient statistical calculations

scores = np.array([72, 85, 91, 68, 77, 88])

print(f"Scores: {scores}")
print(f"Average: {scores.mean():.2f}")
print(f"Highest score: {scores.max()}")
print(f"Lowest score: {scores.min()}")
print(f"Total score: {scores.sum()}")
```
